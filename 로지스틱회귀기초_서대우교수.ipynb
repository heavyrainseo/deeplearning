# 로지스틱회귀

## 공식

$sigmoid = \frac {1}{1+e ^{-(ax+b)}}$

$cost = -(y log h+(1-y)log(1-h))$

차트로 보기 : https://www.desmos.com/calculator/8c6144b9a1

- activation : sigmoid
- loss : binary_crossentropy
- optimizer : sgd

```python
#@title 데이터 로딩
import numpy as np

x = np.array(list(range(2, 15, 2)))
y = np.array([0,0,0,1,1,1,1])
```

```python
#@title 모델 생성
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()
model.add(Input(shape=(1,)))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy', optimizer='sgd')
model.summary()
```

```python
#@title 학습 실행
%%time
history = model.fit(x, y, epochs=2500, verbose=0)
model.get_weights()
```

```python
#@title loss chart
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.show()
```

```python
#@title predict
model.predict(np.array([3,6,9]))
```

## XOR 문제


```python
import numpy as np

x = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,1,1,0])
```

```python
#@title 모델 생성
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()
model.add(Input(shape=(2,)))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy', optimizer='sgd')
model.summary()
```

```python
#@title 학습 실행
%%time
history = model.fit(x, y, epochs=2500, verbose=0)
model.get_weights()
```

```python
#@title loss chart
import matplotlib.pyplot as plt
plt.plot(history.history['loss'])
plt.show()
```

```python
#@title predict
model.predict(np.array([[0,0], [0,1], [1,0], [1,1]]))
```

```python
#@title hidden layer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()
model.add(Input(shape=(2,)))
model.add(Dense(4, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy', optimizer='sgd')
model.summary()
```

```python
#@title 학습 실행
%%time
history = model.fit(x, y, epochs=2500, verbose=0)
model.get_weights()
```

```python
#@title predict
np.round(model.predict(np.array([[0,0], [0,1], [1,0], [1,1]])))
```

# 로지스틱회귀 응용 - geyser

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
gs = sns.load_dataset('geyser')
gs.head()
```

```python
sns.pairplot(gs, hue='kind')
```

```python
print(gs.groupby('kind', observed=False).count())
gs.kind = gs.kind.map({'long':1, 'short':0})
```

```python
gs.info()
```

```python
x = gs[['duration','waiting']]
y = gs['kind']
x, y
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()
model.add(Input(shape=(2,)))
model.add(Dense(16, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
%%time
from tensorflow.keras.callbacks import EarlyStopping

esc = EarlyStopping(patience=20, monitor='loss')
history = model.fit(x, y, validation_split=0.2, epochs=1000, verbose=0, callbacks=[esc])
```

```python
model.evaluate(x, y)
```

```python
plt.figure(figsize=(10,4))
ax1 = plt.subplot(1,2,1)
ax1.plot(history.history['loss'], label='loss')
ax1.plot(history.history['val_loss'], label='val_loss')
ax1.legend()
ax1.grid()

ax2 = plt.subplot(1,2,2)
ax2.plot(history.history['accuracy'], label='acc')
ax2.plot(history.history['val_accuracy'], label='val_acc')
ax2.legend()
ax2.grid()
```

```python
pred = np.round(model.predict(x),0)
res = pd.DataFrame({'y':y, 'pred':pred.flatten()})
res[res.y != res.pred]
```
