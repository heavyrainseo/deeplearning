# Python code

```python
import numpy as np

a, b = np.random.rand(), np.random.rand()
print(a,b)
```

```python
x = np.array([2,4,6,8])
y = np.array([81,93,91,97])
lr = 0.01
mses = []
n = len(x)

for i in range(1000):
    cost_a = -2/n * (x * (y - (a * x +b))).sum()
    cost_b = -2/n * (y-(a * x + b)).sum()
    mse = ((y - (a*x+b))**2).mean()
    mses.append(mse)
    a -= lr * cost_a
    b -= lr * cost_b
    print(a, b)
```

```python
import matplotlib.pyplot as plt

plt.plot(mses)
plt.show()
```

```python
plt.scatter(x, y)
plt.plot(x, a*x+b, 'r--')
plt.show()
```

```python
test_x = np.array([1,2,3,4])
print(a*test_x+b)
```

# Tensorflow code

## Simple LR

```python
import tensorflow as tf

model = tf.keras.models.Sequential()
model.add(tf.keras.Input(shape=(1,)))
model.add(tf.keras.layers.Dense(1, activation="linear"))
model.compile(optimizer="sgd", loss="mse")
print(model.get_weights())
print(model.summary())
```

```python
hist = model.fit(x, y, epochs=1000, verbose=1)
```

```python
import matplotlib.pyplot as plt

print(model.get_weights())

plt.plot(hist.history['loss'])
plt.show()
```

```python
test_x = np.array(range(1,5))
model.predict(test_x)
```

## Multiple LR

```python
import numpy as np

x = np.array([[2,0],[4,4],[6,2],[8,3]])
y = np.array([81,93,91,97])
```

```python
import tensorflow as tf

model = tf.keras.models.Sequential()
model.add(tf.keras.Input(shape=(2,)))
model.add(tf.keras.layers.Dense(1, activation="linear"))
model.compile(optimizer="sgd", loss="mse")
print(model.get_weights())
print(model.summary())
```

```python
hist = model.fit(x, y, epochs=1000, verbose=2)
model.get_weights()
```

```python
import matplotlib.pyplot as plt

plt.plot(hist.history['loss'])
plt.show()
```

```python
test_x = np.array([[0,2],[2,4],[5,0]])
print(model.predict(test_x))
```
